##Experiment 2.1

Write a Python program to perform basic text preprocessing on a given data. The program should:
* Convert text to lowercase. (Also display how many uppercase?)
* Remove punctuation.        (Also display how many different types of punctuations?)
* Remove numbers.            (Also display how many numbers?)
* Remove extra whitespace. (How much extra whitespace is present?)
* Display the cleaned text.


```python
import re
import string
from collections import Counter
```

```python
filename = "/content/drive/MyDrive/CL-NLP/Experiment-2/2.1_text_data.txt"

with open(filename, "r", encoding="utf-8") as file:
    text = file.read()

print("Original Text:")
print("-" * 80)
print(text)
```

```python
uppercase_count = sum(1 for char in text if char.isupper())

lowercase_text = text.lower()

print("Number of uppercase characters:", uppercase_count)

print("\nText after converting to lowercase:")
print("-" * 80)
print(lowercase_text)
```

```python
punctuation_chars = [char for char in text if char in string.punctuation]

different_punctuation = sorted(set(punctuation_chars))

print("Punctuation characters found:")
print(punctuation_chars)

print("\nDifferent types of punctuation:")
print(different_punctuation)

print("\nNumber of different punctuation types:",
      len(different_punctuation))

print("\nTotal punctuation characters:",
      len(punctuation_chars))
```

```python
text_without_punctuation = text.translate(
    str.maketrans("", "", string.punctuation)
)

print("Text after removing punctuation:")
print("-" * 80)
print(text_without_punctuation)
```

```python
numbers = re.findall(r'\d+', text)

print("Numbers found:")
print(numbers)

print("\nNumber of numeric occurrences:", len(numbers))
```

```python
text_without_numbers = re.sub(r'\d+', '', text)

print("Text after removing numbers:")
print("-" * 80)
print(text_without_numbers)
```

```python
total_whitespace = sum(1 for char in text if char.isspace())

normalized_text = re.sub(r'\s+', ' ', text).strip()

normalized_whitespace = normalized_text.count(' ')

extra_whitespace = total_whitespace - normalized_whitespace

print("Total whitespace characters:", total_whitespace)
print("Whitespace after normalization:", normalized_whitespace)
print("Extra whitespace:", extra_whitespace)
```

```python
text_without_extra_whitespace = re.sub(r'\s+', ' ', text).strip()

print("Text after removing extra whitespace:")
print("-" * 80)
print(text_without_extra_whitespace)
```

```python
cleaned_text = text.lower()

cleaned_text = cleaned_text.translate(
    str.maketrans("", "", string.punctuation)
)

cleaned_text = re.sub(r'\d+', '', cleaned_text)

cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()

print("FINAL CLEANED TEXT:")
print("-" * 80)
print(cleaned_text)
```

```python
uppercase_count = sum(1 for char in text if char.isupper())

punctuation_chars = [
    char for char in text
    if char in string.punctuation
]

different_punctuation = sorted(set(punctuation_chars))

numbers = re.findall(r'\d+', text)

total_whitespace = sum(1 for char in text if char.isspace())
normalized_text = re.sub(r'\s+', ' ', text).strip()
normalized_whitespace = normalized_text.count(' ')
extra_whitespace = total_whitespace - normalized_whitespace


print("=" * 60)
print("TEXT PREPROCESSING STATISTICS")
print("=" * 60)

print("1. Uppercase characters        :", uppercase_count)

print("2. Total punctuation characters:", len(punctuation_chars))

print("3. Different punctuation types :", len(different_punctuation))
print("   Punctuation types           :", different_punctuation)

print("4. Numbers found               :", numbers)
print("   Number of numeric occurrences:", len(numbers))

print("5. Total whitespace characters :", total_whitespace)
print("   Extra whitespace             :", extra_whitespace)

print("=" * 60)
```

```python
print("=" * 80)
print("ORIGINAL TEXT")
print("=" * 80)

print(text)

print("\n\n")

print("=" * 80)
print("CLEANED TEXT")
print("=" * 80)

print(cleaned_text)
```

##Experiment 2.2

Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
(Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)


```python
filename = "/content/drive/MyDrive/CL-NLP/Experiment-2/2.2_tokenization_data.txt"

with open(filename, "r", encoding="utf-8") as file:
    text = file.read()

print("Original Text:")
print("-" * 80)
print(text)
```

###NTLK

```python
!pip install -q nltk
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize

sentences_nltk = sent_tokenize(text)

print("=" * 80)
print("NLTK - SENTENCE TOKENIZATION")
print("=" * 80)

for i, sentence in enumerate(sentences_nltk, start=1):
    print(f"Sentence {i}: {sentence}")
```

```python
from nltk.tokenize import word_tokenize

words_nltk = word_tokenize(text)

print("=" * 80)
print("NLTK - WORD TOKENIZATION")
print("=" * 80)

for i, word in enumerate(words_nltk, start=1):
    print(f"{i}. {word}")
```

```python
print("=" * 80)
print("NLTK TOKENIZATION RESULTS")
print("=" * 80)

print("\nSENTENCES:")
for i, sentence in enumerate(sentences_nltk, start=1):
    print(f"{i}. {sentence}")

print("\nWORDS:")
print(words_nltk)

print("\nTotal sentences:", len(sentences_nltk))
print("Total word tokens:", len(words_nltk))
```

###Spacy

```python
import spacy

nlp = spacy.load("en_core_web_sm")

doc = nlp(text)
```

```python
sentences_spacy = list(doc.sents)

print("=" * 80)
print("SPACY - SENTENCE TOKENIZATION")
print("=" * 80)

for i, sentence in enumerate(sentences_spacy, start=1):
    print(f"Sentence {i}: {sentence.text}")
```

```python
words_spacy = [token.text for token in doc]

print("=" * 80)
print("SPACY - WORD TOKENIZATION")
print("=" * 80)

for i, word in enumerate(words_spacy, start=1):
    print(f"{i}. {word}")
```

```python
print("=" * 80)
print("SPACY TOKENIZATION RESULTS")
print("=" * 80)

print("\nSENTENCES:")

for i, sentence in enumerate(sentences_spacy, start=1):
    print(f"{i}. {sentence.text}")

print("\nWORDS:")
print(words_spacy)

print("\nTotal sentences:", len(sentences_spacy))
print("Total word tokens:", len(words_spacy))
```

###Without Libraries

```python
import re

sentences_manual = re.split(r'(?<=[.!?])\s+', text.strip())

print("=" * 80)
print("WITHOUT NLTK/SPACY - SENTENCE TOKENIZATION")
print("=" * 80)

for i, sentence in enumerate(sentences_manual, start=1):
    print(f"Sentence {i}: {sentence}")
```

```python
words_manual = re.findall(r'\b\w+\b|[^\w\s]', text)

print("=" * 80)
print("WITHOUT NLTK/SPACY - WORD TOKENIZATION")
print("=" * 80)

for i, word in enumerate(words_manual, start=1):
    print(f"{i}. {word}")
```

```python
print("=" * 80)
print("MANUAL TOKENIZATION RESULTS")
print("=" * 80)

print("\nSENTENCES:")

for i, sentence in enumerate(sentences_manual, start=1):
    print(f"{i}. {sentence}")

print("\nWORDS:")
print(words_manual)

print("\nTotal sentences:", len(sentences_manual))
print("Total word tokens:", len(words_manual))
```

```python
print("=" * 80)
print("TOKENIZATION COMPARISON")
print("=" * 80)

print("\nNLTK")
print("Number of sentences:", len(sentences_nltk))
print("Number of word tokens:", len(words_nltk))

print("\nspaCy")
print("Number of sentences:", len(sentences_spacy))
print("Number of word tokens:", len(words_spacy))

print("\nWithout NLTK/spaCy")
print("Number of sentences:", len(sentences_manual))
print("Number of word tokens:", len(words_manual))
```

##Experiment 2.3
Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
filename = "/content/drive/MyDrive/CL-NLP/Experiment-2/2.3_clean_data.txt"

with open(filename, "r", encoding="utf-8") as file:
    text = file.read()

print("=" * 80)
print("ORIGINAL TEXT")
print("=" * 80)

print(text)
```

```python
stop_words = {
    "a",
    "an",
    "the",
    "is",
    "are",
    "was",
    "were",
    "in",
    "of",
    "and",
    "to",
    "for",
    "on",
    "with",
    "from",
    "by",
    "this",
    "that",
    "it",
    "as",
    "at",
    "be",
    "can",
    "such",
    "one",
    "into",
    "its",
    "these",
    "those"
}

print("Stop Words:")
print(sorted(stop_words))
```

```python
import re

tokens = re.findall(r'\b\w+\b', text.lower())

print("=" * 80)
print("ORIGINAL TOKENS")
print("=" * 80)

print(tokens)
```

```python
found_stop_words = [
    token for token in tokens
    if token in stop_words
]

print("=" * 80)
print("STOP WORDS FOUND IN TEXT")
print("=" * 80)

print(found_stop_words)

print("\nTotal stop words found:", len(found_stop_words))
```

```python
filtered_tokens = [
    token for token in tokens
    if token not in stop_words
]

print("=" * 80)
print("FILTERED TOKENS")
print("=" * 80)

print(filtered_tokens)

print("\nTotal original tokens:", len(tokens))
print("Total stop words:", len(found_stop_words))
print("Total filtered tokens:", len(filtered_tokens))
```

```python
print("=" * 80)
print("EXPERIMENT 2.3 - STOP WORD REMOVAL")
print("=" * 80)

print("\n1. ORIGINAL TEXT")
print("-" * 80)
print(text)

print("\n2. ORIGINAL TOKENS")
print("-" * 80)
print(tokens)

print("\n3. STOP WORDS")
print("-" * 80)
print(found_stop_words)

print("\n4. FILTERED TOKENS")
print("-" * 80)
print(filtered_tokens)

print("\n5. STATISTICS")
print("-" * 80)
print("Total original tokens :", len(tokens))
print("Stop words removed    :", len(found_stop_words))
print("Remaining tokens      :", len(filtered_tokens))
```

```python
unique_stop_words = sorted(set(found_stop_words))

print("=" * 80)
print("UNIQUE STOP WORDS FOUND")
print("=" * 80)

print(unique_stop_words)

print("\nNumber of different stop words:",
      len(unique_stop_words))
```

```python
from collections import Counter

stop_word_frequency = Counter(found_stop_words)

print("=" * 80)
print("STOP WORD FREQUENCY")
print("=" * 80)

for word, count in stop_word_frequency.most_common():
    print(f"{word:10} : {count}")
```

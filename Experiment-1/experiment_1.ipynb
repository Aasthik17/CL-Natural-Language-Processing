#Experiment-1: Data Analysis

##Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/drive/MyDrive/CL-NLP/Experiment-1/employee_information_100.csv")

avg_salary_by_dept = df.groupby('Department')['Salary'].mean()

plt.figure(figsize=(10, 6))
avg_salary_by_dept.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Average Salary by Department', fontsize=14)
plt.xlabel('Department', fontsize=12)
plt.ylabel('Average Salary', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

##Experiment 1.2: Display the number of employees working in each department.

```python
employee_counts = df['Department'].value_counts()
print(employee_counts)
```

##Experiment 1.3:Display the percentage of male and female employees using a pie chart

```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(8, 8))
gender_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Percentage of Male and Female Employees')
plt.ylabel('')
plt.show()
```

##Experiment 1.4: Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(10, 6))
plt.hist(df['Salary'], bins=20, edgecolor='black')
plt.title('Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

##Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.


```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Experience_Years'], df['Salary'])
plt.title('Relationship Between Experience and Salary')
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.show()
```

##Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top_10_employees = df.nlargest(10, 'Salary')
print(top_10_employees)
```

##Experiment 1.7: Find the highest salary in every department.


```python
highest_salary = df.groupby('Department')['Salary'].max()
print(highest_salary)
```

##Experiment 1.8: Display employees whose salary is greater than the overall average salary.


```python
avg_salary = df['Salary'].mean()
above_avg_employees = df[df['Salary'] > avg_salary]
print(above_avg_employees)
```

##Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_experience = df.groupby('Department')['Experience_Years'].mean()
print(avg_experience)
```

##Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
plt.figure(figsize=(10, 6))
plt.hist(df['Age'], bins=20, edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```
